# 1. Probability & Simulation Problem (Customer Churn)

A company starts with 10 customers (N = 10). Each month, every customer independently has a 30% probability of leaving the service.

1. What is the probability that exactly 1 customer remains after 12 months?
2. Simulate the probability using the Monte Carlo simulation.

### Break it down:

No of customers = 10

Probability of 1 customer leaving each month = 0.3

Probability of 1 customer staying each month = 0.7 

Note: the event of other customers doesn't affect the independent event of another customer leaving/staying

So, the probability of:

1 customer staying for 1 month = 0.7

1 customer staying for 2 months = 0.7 x 0.7 = 0.7^2

1 customer staying for 12 months = 0.7 x 0.7 x 0.7 x ......... = 0.7^12

Solution 1: Exact analytical solution in Python

```python
import math

# Given values
N = 10
p_leave = 0.3
p_stay = 0.7
months = 12

# Probability one customer survives all 12 months
p_survive_12 = p_stay ** months

# Probability exactly 1 customer remains
customer_exact_1 = math.comb(N, 1) * (p_survive_12 ** 1) * ((1 - p_survive_12) ** (N - 1))

print("Probability one customer survives 12 months:", p_survive_12)
print("Probability exactly 1 customer remains after 12 months:", customer_exact_1)
print("Percentage:", customer_exact_1 * 100)
```

Solution 2: Monte Carlo idea

Monte Carlo means:

1. simulate the 10 customers
2. let each customer face 12 monthly survival checks
3. count how many remain after 12 months
4. repeat many times
5. estimate the probability by:

(number of simulations with exactly 1 remaining)/(total simulations)

```python
import numpy as np

# Given values
N = 10
p_leave = 0.3
months = 12
num_simulations = 100_000

results = []

for _ in range(num_simulations):
    customers_remaining = N
    
    for _ in range(months):
        # each remaining customer survives with probability 0.7
        customers_remaining = np.random.binomial(customers_remaining, 1 - p_leave)
        
        if customers_remaining == 0:
            break
    
    results.append(customers_remaining)

results = np.array(results)
estimated_probability = np.mean(results == 1)

print("Monte Carlo estimate:", estimated_probability)
print("Percentage:", estimated_probability * 100)
```
